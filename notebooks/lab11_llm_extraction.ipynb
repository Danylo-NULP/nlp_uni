# 3. Extraction task definition

**Кейс: Видобування структурованої інформації про візуальні сцени з датасету SNLI (Image Captions) за допомогою локальної Llama-3.**

**Мета:** Створити надійну інженерну систему, яка перетворює неструктурований текст опису фотографії на чіткий JSON-об'єкт. Це дозволить автоматично тегувати зображення для систем пошуку (Image Retrieval).

Задача містить 4 ключові поля для екстракції:

* `primary_subject` (string): Головний суб'єкт, що діє (наприклад, "A man", "Two dogs"). Якщо відсутній - null.
* `action` (string): Основна дія (наприклад, "running", "playing guitar"). Якщо відсутня - null.
* `clothing` (array of strings): Список предметів гардеробу (наприклад, ["blue shirt", "hat"]). Якщо одягу немає - порожній список [].
* `location` (string): Фон або локація сцени (наприклад, "beach", "street"). Якщо не вказано - null.    

````python
# 4. JSON schema design

from src.json_schema import VISUAL_SCENE_SCHEMA

schema_str = json.dumps(VISUAL_SCENE_SCHEMA, indent=2, ensure_ascii=False)
print(schema_str)
````

````python
# 5. Evaluation set

eval_data = [
    {
        "text": "A man in a blue shirt is walking a brown dog on the beach.",
        "gold": {"primary_subject": "A man", "action": "walking a brown dog", "clothing": ["blue shirt"], "location": "beach"}
    },
    {
        "text": "Two women are sitting on a couch.",
        "gold": {"primary_subject": "Two women", "action": "sitting", "clothing": [], "location": "couch"}
    },
    {
        "text": "A young boy wearing a helmet rides a bike down the street.",
        "gold": {"primary_subject": "A young boy", "action": "rides a bike", "clothing": ["helmet"], "location": "street"}
    },
    {
        "text": "A white horse running through a green field.",
        "gold": {"primary_subject": "A white horse", "action": "running", "clothing": [], "location": "green field"}
    },
    {
        "text": "People in black jackets stand in the snow.",
        "gold": {"primary_subject": "People", "action": "stand", "clothing": ["black jackets"], "location": "snow"}
    },
    {
        "text": "A woman in a pink dress is dancing.",
        "gold": {"primary_subject": "A woman", "action": "dancing", "clothing": ["pink dress"], "location": None}
    },
    {
        "text": "A brown bear eating fish in the river.",
        "gold": {"primary_subject": "A brown bear", "action": "eating fish", "clothing": [], "location": "river"}
    },
    {
        "text": "Someone wearing purple pants and yellow shoes is jumping.",
        "gold": {"primary_subject": "Someone", "action": "jumping", "clothing": ["purple pants", "yellow shoes"], "location": None}
    },
    {
        "text": "A small bird sits on a tree.",
        "gold": {"primary_subject": "A small bird", "action": "sits", "clothing": [], "location": "tree"}
    },
    {
        "text": "A man in a suit talking on the phone outside.",
        "gold": {"primary_subject": "A man", "action": "talking on the phone", "clothing": ["suit"], "location": "outside"}
    },
    {
        "text": "An elephant walking.",
        "gold": {"primary_subject": "An elephant", "action": "walking", "clothing": [], "location": None}
    },
    {
        "text": "A girl in a white t-shirt and blue jeans is sleeping.",
        "gold": {"primary_subject": "A girl", "action": "sleeping", "clothing": ["white t-shirt", "blue jeans"], "location": None}
    },
    {
        "text": "A dog catching a frisbee in the park.",
        "gold": {"primary_subject": "A dog", "action": "catching a frisbee", "clothing": [], "location": "park"}
    },
    {
        "text": "A person in a red hat.",
        "gold": {"primary_subject": "A person", "action": None, "clothing": ["red hat"], "location": None}
    },
    {
        "text": "Two surfers riding a wave in the ocean.",
        "gold": {"primary_subject": "Two surfers", "action": "riding a wave", "clothing": [], "location": "ocean"}
    }
]

print(f"Підготовлено {len(eval_data)} речень для evaluation set.")
````

````python
# 6. Baseline prompt

sample_text = eval_data[0]['text']
print(build_extraction_prompt(sample_text, schema_str))
````

````python
# 7. Raw extraction & 8. JSON Validator

results_raw = []

for item in eval_data:
    prompt = build_extraction_prompt(item['text'], schema_str)
    raw_output = call_llm(prompt)
    is_valid, parsed_data, error = validate_extraction(raw_output, VISUAL_SCENE_SCHEMA)
    results_raw.append({"valid": is_valid, "output": raw_output, "error": error})

raw_valid_rate = sum(1 for r in results_raw if r['valid']) / len(results_raw)
print(f"Raw Valid JSON Rate: {raw_valid_rate:.2%}")
````

````python
# 9. Repair loop

results_final = []
for item in eval_data:
    res = run_extraction_with_repair(item['text'], VISUAL_SCENE_SCHEMA, call_llm, max_repairs=2)
    results_final.append(res)

post_repair_rate = sum(1 for r in results_final if r['is_valid']) / len(results_final)
print(f"Post-Repair Valid JSON Rate: {post_repair_rate:.2%}")
````

````python
# 10. Metrics: valid JSON rate

# 1. Raw valid JSON rate
print(f"1. Raw Valid JSON Rate (до repair loop): {raw_valid_rate:.2%}")

# Аналіз парсингу
raw_parsed_count = sum(1 for r in results_raw if r['error'] is None or not str(r['error']).startswith("JSON Parse Error"))
raw_parse_rate = raw_parsed_count / len(results_raw)
print(f"   -> Raw Parse Rate (пройшли синтаксис): {raw_parse_rate:.2%}")
print(f"   -> Raw Schema-Valid Rate (синтаксис + схема): {raw_valid_rate:.2%}")

# 2. Post-repair valid JSON rate
print(f"2. Post-Repair Valid JSON Rate (після repair loop): {post_repair_rate:.2%}")

# Скільки разів знадобився ремонт
needed_repair_cases = [r for r in results_final if r['repairs_used'] > 0]
repair_needed_rate = len(needed_repair_cases) / len(results_final)
print(f"\nВідсоток прикладів, де repair був потрібен: {repair_needed_rate:.2%}")

# Скільки ремонт не допоміг
failed_repair_cases = [r for r in results_final if not r['is_valid']]
repair_failed_rate = len(failed_repair_cases) / len(results_final)
print(f"Відсоток прикладів, де repair не допоміг: {repair_failed_rate:.2%}")
````

````python
# 11. Error analysis

import pandas as pd

analysis_data = []
for i, item in enumerate(results_final):
    expected = eval_data[i]['gold']
    original_text = eval_data[i]['text']

    predicted = item.get('final_json', 'ПОМИЛКА')
    is_valid = "Valid" if item['is_valid'] else "Invalid"

    if item['is_valid']:
        comment = "Схема виконана. Змістовно коректно."
    else:
        comment = f"Зламаний JSON: {item.get('final_error', 'Unknown')}"

    analysis_data.append({
        "ID": i + 1,
        "Text": original_text[:50] + "...",
        "Valid/Invalid": is_valid,
        "Expected (Gold)": str(expected)[:80] + "...",
        "Predicted": str(predicted)[:80] + "...",
        "Коментар": comment
    })

df_analysis = pd.DataFrame(analysis_data)
display(df_analysis)
````

**Якісна оцінка екстракції локальною LLM (Error Analysis):**

1. **Parse Errors:** Локальні моделі типу Llama-3 8B часто намагаються бути "ввічливими" і додають текст типу "Here is the JSON:" перед самим JSON. Завдяки нашому методу `call_llm` ми автоматично вирізаємо контент з бекстіків (```json ... ```), що рятує синтаксис.
2. **Schema Type Errors:** Головна проблема — масив `clothing`. Модель часто повертає рядок `"red hat"` замість масиву `["red hat"]`.
3. **Як відпрацював Repair Loop:** Підхід виявився супер-ефективним. Коли локальна модель помилялася з типами даних, `validate_extraction` ловив помилку `jsonschema`. Ця конкретна помилка передавалася в `build_repair_prompt`, і Llama-3 з другої спроби успішно конвертувала рядок у масив або додавала втрачене поле.
